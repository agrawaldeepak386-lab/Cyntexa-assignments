##5. Configure M2M (service principal) authentication for the CLI and use it instead of your personal U2M login for a deploy command.


## Configure M2M (Service Principal) Authentication for Databricks CLI

### Objective

Configure **OAuth M2M (Machine-to-Machine)** authentication using a Databricks **Service Principal** and use it instead of personal **U2M (User-to-Machine)** authentication for bundle deployment.

### Authentication flow

```
Service Principal
    ↓
OAuth M2M
    ↓
Databricks CLI
    ↓
Databricks Workspace
    ↓
Bundle Deployment
```
1. **Create a Service Principal**

A Service Principal is used for non-human authentication.

Create the Service Principal using the Databricks CLI:

```
databricks service-principals create --display-name "deepak1"
```

Example response:
```
{
"active": true,
"applicationId": "7ace7004-ebbd-487c-ad03-6d74badcd6c6",
"displayName": "deepak1",
"id": "74781335647560"
}
```
The important value for OAuth M2M authentication is the:

Application ID

```
7ace7004-ebbd-487c-ad03-6d74badcd6c6
```

2. Create an OAuth Secret

Create an OAuth secret for the Service Principal.

First check the available command in the installed Databricks CLI:

```
databricks service-principal-secrets-proxy --help
```

Create the secret:
```
databricks service-principal-secrets-proxy create 74781335647560
```
The response contains a secret similar to:
```
{
"create_time": "2026-09-10T09:54:13.000Z",
"expire_time": "2028-09-09T09:54:13Z",
"id": "<SECRET_ID>",
"secret": "<OAUTH_SECRET>",
"status": "ACTIVE"
}
```

**Important: The OAuth secret is sensitive. Do not commit it to Git, put it inside databricks.yml, or share it publicly.**

3. Configure the Databricks CLI Profile

The Databricks CLI stores authentication profiles in:

```
$HOME.databrickscfg
```

On Windows PowerShell, open the file using:

```
notepad $HOME.databrickscfg
```
Add an M2M profile:

```
[MY_SP]
host = https://dbc-23ec29fd-2d98.cloud.databricks.com
client_id = 7ace7004-ebbd-487c-ad03-6d74badcd6c6
client_secret = <OAUTH_SECRET>
```

Replace <OAUTH_SECRET> with the newly generated OAuth secret.

4. Verify the M2M Profile

List the configured authentication profiles:

```
databricks auth profiles
```

Test authentication using the Service Principal:

```
databricks current-user me -p MY_SP
```

The command should authenticate as the Service Principal rather than the personal Databricks user.

5. Configure the Bundle Target

For the target that should use the Service Principal, configure the profile in databricks.yml.

Example:

```
targets:
 dev:
  mode: development
 workspace:
  host: https://dbc-23ec29fd-2d98.cloud.databricks.com
  profile: MY_SP
 variables:
  catalog: dev
  schema: dev
```

The profile setting tells the Databricks CLI to use the M2M authentication profile instead of personal U2M authentication.

6. Validate the Bundle

From the bundle project directory, run:

```
databricks bundle validate -t dev
```

If validation succeeds, the bundle is ready for deployment.

7. Deploy Using M2M Authentication

Deploy the bundle:

```
databricks bundle deploy -t dev
```

The deployment now uses:

```
Databricks CLI
↓
MY_SP profile
↓
OAuth M2M
↓
Service Principal
↓
Databricks Workspace
↓
Bundle Deployment
```

The CLI is no longer using the personal U2M identity for this deployment.

8. Verify the Deployment Identity

To verify which identity is being used:

```
databricks current-user me -p MY_SP
```

The returned identity should correspond to the Service Principal.

9. Important Difference: Deployment Identity vs Job Runtime Identity

M2M authentication controls who the CLI authenticates as during deployment.

It is different from the identity that executes a Databricks job.

CLI authentication

```
workspace:
profile: MY_SP
```

This means:

Deploy the bundle using the Service Principal authentication profile.

Job runtime identity

If the job itself should run as a Service Principal, configure:

```
run_as:
service_principal_name: "<SERVICE_PRINCIPAL_APPLICATION_ID>"
```

Therefore:

```
workspace.profile
↓
Who deploys the bundle

run_as
↓
Who runs the job
```

10. DEV and PROD with Different Service Principals

For a real project, different authentication profiles can be configured for different environments.

Example:

```
[DEV_SP]
host = https://dbc-23ec29fd-2d98.cloud.databricks.com
client_id = <DEV_SP_APPLICATION_ID>
client_secret = <DEV_SP_SECRET>

[PROD_SP]
host = https://<FRIEND_WORKSPACE>.cloud.databricks.com
client_id = <PROD_SP_APPLICATION_ID>
client_secret = <PROD_SP_SECRET>
```

The bundle can then use different profiles:

```
targets:

dev:
 mode: development
 workspace:
  host: https://dbc-23ec29fd-2d98.cloud.databricks.com
  profile: DEV_SP
 variables:
  catalog: dev
  schema: dev

prod:
 mode: production
 workspace:
  host: https://<FRIEND_WORKSPACE>.cloud.databricks.com
  profile: PROD_SP
 variables:
  catalog: dev
  schema: prod
```

Deploy DEV:

```
databricks bundle deploy -t dev
```

Deploy PROD:

```
databricks bundle deploy -t prod
```

11. Permissions Required

Creating a Service Principal does not automatically give it permission to deploy everywhere.

The Service Principal must have the required permissions in the target workspace, including access to:

Workspace folders
Bundle deployment location
Jobs and other resources being deployed
Catalogs and schemas if the deployment requires data access
Any other resources used by the bundle

For a friend's PROD workspace, the Service Principal must be authorized in that workspace.

12. Security Best Practices
Never commit the OAuth secret

Do not put this in databricks.yml:

```
client_secret: "<SECRET>"
```

Do not commit .databrickscfg to Git.

Use a dedicated Service Principal for production

Instead of using a personal user identity:

```
Developer User
↓
Production Deployment


prefer:


Production Service Principal
    ↓
Production Deployment
```

### Rotate secrets

OAuth secrets have an expiration date. Generate a new secret before the current secret expires and update the CLI configuration.

## Final Result

After configuration, the deployment architecture is:

```
                    Git Repository
                          │
                          ▼
                   databricks.yml
                          │
                          ▼
                    Databricks CLI
                          │
                    OAuth M2M
                          │
                          ▼
                 Service Principal
                          │
                          ▼
                 Databricks Workspace
                          │
                          ▼
                    Bundle Deploy
```

This provides an **unattended, non-personal authentication mechanism** for Databricks
---

### Comparison: U2M vs M2M Authentication

| Feature | U2M (User-to-Machine) | M2M (Machine-to-Machine) |
|---------|----------------------|-------------------------|
| **Use Case** | Interactive development, personal workflows | CI/CD pipelines, automation, production |
| **Identity** | Personal user account | Service principal |
| **Authentication** | OAuth with browser login | Client ID + Secret |
| **Token Lifetime** | Session-based, requires refresh | Long-lived, automated |
| **Permissions** | User's personal permissions | Service principal permissions |
| **Audit Trail** | Tied to user | Tied to service principal |
| **Best For** | Development, exploration | Production deployments, automation |

1. **Trigger**: Runs on every pull request to `main`, `dev`, or `staging` branches

3. **Validation**: Executes `databricks bundle validate -t dev` without deploying

##7. Extend the GitHub Actions workflow to deploy to prod on merge to main using OIDC authentication (no stored secrets), including the correct permissions: id-token: write block.

## GitHub Actions Workflow with OIDC Deployment to Production

### What is OIDC and Why Use It?

**OIDC (OpenID Connect)** lets GitHub Actions authenticate to Databricks **without storing any secrets** (no PAT, no OAuth client secret). Instead, GitHub issues a short-lived OIDC JWT token per workflow run, which Databricks exchanges for a temporary access token.

```
GitHub Actions run (merge to main)
        |
        |  1. Request OIDC JWT (signed by GitHub)
        v
GitHub OIDC Provider (token.actions.githubusercontent.com)
        |
        |  2. Present JWT to Databricks
        v
Databricks OIDC Token Endpoint
        |
        |  3. Validate JWT against trust policy
        |  4. Issue short-lived Databricks access token
        v
Databricks CLI deploys bundle to prod
```

**Key security advantage:** No long-lived credentials in GitHub Secrets. The workspace URL (`DATABRICKS_HOST`) is public information, not a secret.

---

### Workflow File: `.github/workflows/bundle-deploy-prod.yml`

```yaml
name: Deploy to Production

on:
  push:
    branches:
      - main

jobs:
  deploy-prod:
    name: Deploy Bundle to Production
    runs-on: ubuntu-latest

    # CRITICAL: id-token:write is required for OIDC token exchange
    # contents:read is needed for actions/checkout
    permissions:
      id-token: write   # Allows GitHub to issue OIDC tokens
      contents: read    # Allows reading repository code

    steps:
      - name: Checkout repository
        uses: actions/checkout@v4

      - name: Install Databricks CLI
        uses: databricks/setup-cli@main

      - name: Authenticate to Databricks using OIDC
        uses: databricks/authenticate-oidc@v1
        with:
          # Only the workspace URL is needed - NOT a secret
          # OIDC handles the token exchange automatically
          databricks-host: https://dbc-2bd5e4bd-405a.cloud.databricks.com

      - name: Validate bundle for production
        working-directory: ./Day_9_Assignment/Day_9_vs
        run: databricks bundle validate -t prod

      - name: Deploy to production
        working-directory: ./Day_9_Assignment/Day_9_vs
        run: databricks bundle deploy -t prod

      - name: Post deployment summary
        if: always()
        run: |
          echo "### Production Deployment Complete" >> $GITHUB_STEP_SUMMARY
          echo "" >> $GITHUB_STEP_SUMMARY
          echo "| Field | Value |" >> $GITHUB_STEP_SUMMARY
          echo "|-------|-------|" >> $GITHUB_STEP_SUMMARY
          echo "| Environment | prod |" >> $GITHUB_STEP_SUMMARY
          echo "| Workspace | https://dbc-2bd5e4bd-405a.cloud.databricks.com |" >> $GITHUB_STEP_SUMMARY
          echo "| Commit | ${{ github.sha }} |" >> $GITHUB_STEP_SUMMARY
          echo "| Triggered by | ${{ github.actor }} |" >> $GITHUB_STEP_SUMMARY
          echo "| Auth method | OIDC (no stored secrets) |" >> $GITHUB_STEP_SUMMARY
```

---

### Understanding the `permissions: id-token: write` Block

```yaml
permissions:
  id-token: write   # Mandatory for OIDC - without this, GitHub cannot issue OIDC tokens
  contents: read    # Required for actions/checkout to read repo files
```

**Why `id-token: write` is critical:**
- GitHub Actions generates an OIDC JWT token for every job run
- Without `id-token: write`, the token endpoint is blocked and OIDC authentication fails
- This permission must be set at the **job level** (not workflow level) to limit scope
- Only the deploy job needs it; the validate-on-PR job does NOT need it

**Security scoping:**
- Set `id-token: write` ONLY on the production deploy job, never globally
- This prevents unrelated jobs from requesting OIDC tokens
- GitHub's default permission is `id-token: none` (most restrictive)

---

### How `databricks/authenticate-oidc@v1` Works (Internal Flow)

The official action handles the entire OIDC exchange automatically:

```
Step 1: Action requests GitHub OIDC JWT
   ├─ Audience: databricks
   ├─ Signed by: token.actions.githubusercontent.com
   └─ Lifetime: ~5 minutes (ephemeral)

Step 2: Action sends JWT to Databricks token endpoint
   ├─ POST https://<workspace-host>/oidc/v1/token
   ├─ grant_type=urn:ietf:params:oauth:grant-type:token-exchange
   ├─ subject_token=<GitHub OIDC JWT>
   └─ subject_token_type=urn:ietf:params:oauth:token-type:id_token

Step 3: Databricks validates the JWT
   ├─ Verifies signature against GitHub's JWKS
   ├─ Checks issuer = https://token.actions.githubusercontent.com
   ├─ Checks audience = databricks
   └─ Checks subject claim against trust policy (repo + branch filter)

Step 4: Databricks issues a short-lived access token
   ├─ Lifetime: ~1 hour (ephemeral)
   ├─ Set as DATABRICKS_TOKEN env var automatically
   └─ Used by databricks bundle deploy command
```

No secrets are stored anywhere. The tokens are ephemeral and unique per workflow run.

---

### Databricks Free Edition Limitation (Important)

**OIDC does NOT work on Databricks Free Edition / Community Edition.**

The official Databricks documentation lists these Free Edition restrictions:

> *No access to the account console or account-level APIs.*
> *Authentication is limited to email OTP, Sign in with Google, and Sign in with Microsoft. No SSO or SCIM support.*

OIDC requires:
* **Account console access** to create federation policies (`databricks account service-principal-federation-policy create`)
* **Account-level APIs** to register GitHub as an OIDC identity provider
* **Service principal federation** which is an account-level feature

Since Free Edition blocks account console and account-level APIs, **OIDC cannot be configured**.

##8. Design a rollback plan: if a bundle deploy to prod breaks a job, what CLI commands would you run to redeploy the previous working version quickly?

## Rollback Plan: Restoring Previous Working Bundle Version

### Scenario

A `databricks bundle deploy -t prod` has been executed and the deployment succeeded, but the deployed job is now broken (fails to run, produces incorrect results, or has configuration errors). The objective is to quickly restore the previous working version.

---

### Rollback Strategy Overview

```
Git Repository (source of truth)
        ↓
Rollback to previous commit/tag
        ↓
Redeploy from working version
        ↓
Production workspace restored
```

**Key principle:** Databricks bundles are **stateless and declarative**. The workspace reflects whatever version you deploy from your Git repository. To roll back, simply redeploy from the previous working commit.

---

### Method 1: Git-Based Rollback (Recommended)

#### Prerequisites
- Git repository with commit history
- Previous working version committed and tagged (best practice)

#### Step-by-Step Commands

**1. Identify the last working commit:**

```bash
# View recent commit history
git log --oneline -10

# Example output:
abcd123 (HEAD -> main) Fix: Update job configuration  ← Current (broken)
789efgh Add new transformation step
456def1 Update production parameters            ← Last known good
123abc4 Initial production deploy
```

**2. Check out the previous working version:**

```bash
# Option A: Checkout by commit hash
git checkout 456def1

# Option B: If you tagged releases (best practice)
git checkout v1.2.3

# Option C: Revert to previous commit on main
git revert abcd123  # Creates a new commit that undoes the bad change
git revert 789efgh
```

**3. Validate the bundle configuration:**

```bash
databricks bundle validate -t prod
```

**Expected output:**
```
Target: prod
Workspace:
  Host: https://dbc-2bd5e4bd-405a.cloud.databricks.com
  User: service-principal-name
  Path: /Workspace/.bundle/project_name/prod

✓ Validation OK!
```

**4. Redeploy the working version:**

```bash
databricks bundle deploy -t prod
```

**Expected output:**
```
✓ Uploading bundle files...
✓ Deploying resources...
✓ Updating deployment state...
✓ Deployment complete!
```

**5. Verify the job is restored:**

```bash
# List jobs to confirm deployment
databricks jobs list | grep <job_name>

# Get job configuration
databricks jobs get --job-id <job-id>

# Trigger a test run
databricks jobs run-now --job-id <job-id>

# Monitor the run
databricks runs get --run-id <run-id>
```

**6. Return to main branch (after confirming rollback):**

```bash
# If you used git checkout, return to main
git checkout main

# Now fix the issue in a new branch
git checkout -b fix/broken-production-deploy
```

---

### Method 2: Deploy from Previous Git Tag

If you follow a tagging strategy for production releases:

```bash
# List recent tags
git tag --sort=-creatordate | head -10

# Output example:
v1.2.4  ← Current (broken)
v1.2.3  ← Last known good
v1.2.2
v1.2.1

# Checkout the previous version
git checkout v1.2.3

# Validate and deploy
databricks bundle validate -t prod
databricks bundle deploy -t prod

# After confirming rollback, return to main
git checkout main
```

### Method 3: Emergency Rollback Using Bundle Destroy + Redeploy

**⚠️ Use with extreme caution** - this will delete all deployed resources.

```bash
# Step 1: Destroy the broken deployment (DESTRUCTIVE)
databricks bundle destroy -t prod

# Step 2: Checkout working version
git checkout <previous-working-commit>

# Step 3: Redeploy from clean state
databricks bundle deploy -t prod
```

**Warning:** `databricks bundle destroy` will:
- Delete all jobs defined in the bundle
- Remove all deployed resources
- Clear the deployment state

Only use this if the deployed resources are in an inconsistent state that prevents a normal redeploy.

### Best Practices for Fast Rollbacks

#### 1. **Git Tagging Strategy**

Tag every production deployment:

```bash
# Before deploying to prod
git tag -a v1.2.4 -m "Production release 1.2.4 - Job config update"
git push origin v1.2.4

databricks bundle deploy -t prod
```

This makes rollback trivial:
```bash
git checkout v1.2.3
databricks bundle deploy -t prod
```

#### 2. **Maintain a Deployment Log**

Keep a record of what was deployed when:

```bash
# Create deployment_log.md
echo "## $(date): Deployed commit $(git rev-parse --short HEAD)" >> deployment_log.md
git add deployment_log.md
git commit -m "Log deployment"
```

#### 3. **Smoke Test After Deploy**

Always verify immediately after deployment:

```bash
# In your deployment script
databricks bundle deploy -t prod

# Immediate verification
JOB_ID=$(databricks jobs list | grep my_job | awk '{print $1}')
databricks jobs run-now --job-id $JOB_ID

# Wait and check status
RUN_ID=$(databricks runs list --job-id $JOB_ID --limit 1 | grep RUNNING | awk '{print $1}')
databricks runs get --run-id $RUN_ID

# If failed, immediate rollback
if [ $? -ne 0 ]; then
  echo "Deployment verification failed, rolling back..."
  git checkout v1.2.3
  databricks bundle deploy -t prod
fi
```

#### 4. **Bundle Deployment State**

Databricks stores deployment metadata in the workspace at:
```
/Workspace/.bundle/<bundle-name>/<target>/.internal/
```

This state is automatically updated on each deploy. You don't need to manually manage it—just redeploy from the working Git commit.

#### 5. **Test in Staging First**

Always deploy to staging before production:

```bash
# Deploy to staging
databricks bundle deploy -t staging

# Run integration tests
databricks jobs run-now --job-id <staging-job-id>

# Verify success, then promote to prod
git tag v1.2.4
databricks bundle deploy -t prod
```

### CI/CD Integration (GitHub Actions Example)

Automate rollback detection:

```yaml
name: Verify Production Deployment

on:
  workflow_run:
    workflows: ["Deploy to Production"]
    types: [completed]

jobs:
  verify-and-rollback:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout
        uses: actions/checkout@v4
        with:
          fetch-depth: 0  # Full history for rollback

      - name: Install Databricks CLI
        uses: databricks/setup-cli@main

      - name: Authenticate
        uses: databricks/authenticate-oidc@v1
        with:
          databricks-host: ${{ secrets.PROD_WORKSPACE_URL }}

      - name: Get deployed job
        id: job
        run: |
          JOB_ID=$(databricks jobs list | grep my_production_job | awk '{print $1}')
          echo "job_id=$JOB_ID" >> $GITHUB_OUTPUT

      - name: Trigger smoke test
        id: test
        run: |
          RUN_ID=$(databricks jobs run-now --job-id ${{ steps.job.outputs.job_id }} | jq -r .run_id)
          echo "run_id=$RUN_ID" >> $GITHUB_OUTPUT

      - name: Wait for completion
        run: |
          databricks runs wait --run-id ${{ steps.test.outputs.run_id }}

      - name: Check run status
        id: status
        run: |
          STATUS=$(databricks runs get --run-id ${{ steps.test.outputs.run_id }} | jq -r .state.result_state)
          echo "status=$STATUS" >> $GITHUB_OUTPUT

      - name: Rollback on failure
        if: steps.status.outputs.status != 'SUCCESS'
        run: |
          echo "Deployment verification failed, initiating rollback..."
          PREVIOUS_TAG=$(git describe --tags --abbrev=0 HEAD^)
          git checkout $PREVIOUS_TAG
          databricks bundle validate -t prod
          databricks bundle deploy -t prod
          echo "Rollback to $PREVIOUS_TAG complete"
```

## Stage 3: CI/CD Pipeline

### What Happens After Merge to `main`

**GitHub Actions automatically triggers:**

```yaml
# Simplified workflow view
name: Deploy to Production

on:
  push:
    branches: [main]

jobs:
  deploy:
    steps:
      - Authenticate using OIDC (no secrets stored!)
      - databricks bundle validate -t prod
      - databricks bundle deploy -t prod
```

**The deploy command:**
- Uploads your notebooks and code to the workspace
- Creates or updates jobs, pipelines, and other resources
- Applies production configuration (cluster size, schedules, etc.)
- **Does NOT automatically run jobs** — deployment and execution are separate

## Stage 4: Bundle Lifecycle in the Workspace

### What `databricks bundle deploy` Does

```
Local files          →  Upload to workspace    →  Update resources
├─ databricks.yml        ├─ /Workspace/.bundle/     ├─ Jobs
├─ notebooks/            │    my_project/prod/      ├─ Pipelines  
└─ src/                  └─ Code and state         └─ Serving endpoints
```

**Idempotent**: Running deploy again updates only what changed.

**Three core commands:**

1. **`databricks bundle validate`** — Check configuration (safe, no changes)
2. **`databricks bundle deploy`** — Upload and update resources
3. **`databricks bundle run`** — Trigger a job execution (optional, manual or scheduled)

## Authentication: OIDC (No Secrets!)

**Why OIDC?**
- GitHub issues a **short-lived token** for each workflow run
- Databricks validates the token and issues temporary access
- **No PAT or OAuth secrets** stored in GitHub
- Tokens expire automatically after ~1 hour

**Required in workflow:**
```yaml
permissions:
  id-token: write   # Critical: allows GitHub to issue OIDC tokens
  contents: read
```
